## 5. Experiment to improve (hyperparameter tuning)

A model's first evaluation metrics aren't always its last. One way to improve a models predictions is with hyperparameter tuning.

```python
# Try different amounts of n_estimators (n_estimators is a hyperparameter you can change)
np.random.seed(42)
for i in range(10, 100, 10):  # this will loop through 10, 20, 30, ..., 90
    print(f"Trying model with {i} estimators...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)
    print(f"Model accruacy on test set: {model.score(X_test, y_test) * 100:.2f}%") # display accuracy as a percentage limited to 2 decimal places. the :.2f means 2 decimal places and the % means percentage. it is a shortcut in python and it is similar to doing round(model.score(X_test, y_test) * 100, 2)
```

**Note:** It's best practice to test different hyperparameters with a validation set or cross-validation.

```python
from sklearn.model_selection import cross_val_score

# Try different numbers of estimators with cross-validation and no cross-validation
np.random.seed(42)
for i in range(10, 100, 10):
    print(f"Trying model with {i} estimators...")
    model = RandomForestClassifier(n_estimators=i).fit(X_train, y_train)
    print(f"Model accruacy on test set: {model.score(X_test, y_test)}")
    print(f"Cross-validation score: {np.mean(cross_val_score(model, X, y, cv=5)) * 100}%")
    print("")
```
